```python
from datetime import datetime
from zoneinfo import ZoneInfo

# A naive datetime (no timezone attached)
naive_dt = datetime.now()

# Attach a specific time zone to make it "aware"
london_dt = naive_dt.replace(tzinfo=ZoneInfo("Europe/London"))
london_dt
```

```python
df['temp'] = naive_dt
```

```python
df['london_date_time']=london_dt
```
